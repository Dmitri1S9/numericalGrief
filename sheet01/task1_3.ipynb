## 1.3

1. On the x-axis we have $\log h$, on the y-axis $\log(Ch^\alpha) = \log C + \alpha\log h = C' + \alpha\log h$,
   so $y$ depends linearly on $x$ → straight line.

2. The slope is $\alpha$.

3. $N \mapsto Ce^{-bN}$: on the x-axis $N$, on the y-axis $\log C - bN = C' - bN$,
   so $y$ depends linearly on $x$ with slope $-b$. Only the y-axis is logarithmic → **semilogy**.

4. $f(h) = Ce^{-b/h}$: $\log f = \log C - b/h = C' - b\,h^{-1}$,
   so on the x-axis we take $1/h$ to get a straight line (linear dependence) → **semilogy** vs. $1/h$.

5. $f(h) = Ce^{-b/h^2}$: analogously $\log f = C' - b\,h^{-2}$,
   so on the x-axis we take $1/h^2$ → **semilogy** vs. $1/h^2$ gives a straight line.
   If we plot vs. $h$ instead, we get a hyperbola-like curve, not a straight line.
