```python
import numpy as np
from numpy.polynomial import Polynomial

a = np.array([1, 2, -1, 0, 0, 0, 0, 0])
# x_i = 2Pi*i/n 
# z_i = e**(-ix) - point 
b = np.array([2, -1, 0, 1, 0, 0, 0, 0])

p1 = Polynomial(a)
p2 = Polynomial(b)

Fa = np.fft.fft(a) # Fn a = FFT = DFT
Fb = np.fft.fft(b)      
c  = np.fft.ifft(Fa * Fb) # from a) ifft(Fa * Fb) ifft(F(a*b))

# F̀n(Fn(a * b)) = a * b
# F̀ is inverse F: ifft = F̀ u

print(Fa.round(3), Fb.round(3), c.round(3), sep="\n\n")
print(p1 * p2)
```
