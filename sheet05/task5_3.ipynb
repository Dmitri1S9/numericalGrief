```python
f = lambda x: 1 - x - 4*x**3 + 2*x**5
I = 1104 
a = -2
b =  4
```

```python
wTrapezoidal = np.array([1, 1]) / 2
wSimpson     = np.array([1, 4, 1]) / 6
wMilne       = np.array([7, 32, 12, 32, 7]) / 90

def midpoint(f, a, b):
    return (b - a) * f((a + b) / 2)

def newtonCote(f, a, b, weights):
    splines = np.linspace(a, b, len(weights))
    return (b - a) * np.sum(weights * f(splines))

def compNewtonCote(f, N, a, b, weights):
    h_i = np.linspace(a, b, N + 1)
    return np.sum([newtonCote(f, h_i[i], h_i[i + 1], weights) for i in range(N)])
```

```python
Qa = newtonCote(f, a, b, wTrapezoidal)
Qb = midpoint(f, a, b)
Qc = newtonCote(f, a, b, wSimpson)
Qd = newtonCote(f, a, b, wMilne)
Qe = compNewtonCote(f, 4, a, b, wTrapezoidal)
Qf = compNewtonCote(f, 3, a, b, wSimpson)       
```

```python
print("I =", I)
for name, Q in zip("abcdef", [Qa, Qb, Qc, Qd, Qe, Qf]):
    print(f"({name})  Q = {Q:>10.4f}   rel. error = {abs(I - Q) / abs(I)}")
```
